```python
daily_sales = (
    df.groupby('Order_Date')['Sales_Amount'].sum().sort_index()
)
```

```python
daily_sales
```

```python
plt.figure(figsize=(10,8))

plt.plot(
    daily_sales.index,
    daily_sales.values
)





















plt.title("Daily sales")
plt.xlabel("order_date")
plt.ylabel("total sales amount")


plt.tight_layout()
plt.show()
```

```python
df.columns
```

```python
plt.figure(figsize=(5,6))

sns.histplot(
    data=df,
    x="Sales_Amount",
    bins=30
)

plt.title("Distibution of sales amount")
plt.xlabel("sales_amount")
plt.ylabel("no of order lines")


plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,6))

sns.histplot(
    data=df,
    x="Product_Category",
    bins=3
)

plt.title("Distibution of brand")
plt.xlabel("brand")
plt.ylabel("no of order lines")


plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,6))

category_sales = df.groupby('Product_Category')['Sales_Amount'].sum().sort_values(ascending = False)

sns.barplot(
    x= category_sales.index,
    y=category_sales.values
)

plt.title("Distibution of g")
plt.xlabel("g")
plt.ylabel("no of order lines")


plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,6))

category_sales = df.groupby('Product_Category')['Sales_Amount'].sum().sort_values(ascending = True)

sns.barplot(
    x= category_sales.index,
    y=category_sales.values
)

plt.title("Distibution of g")
plt.xlabel("g")
plt.ylabel("no of order lines")


plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,8))

category_sales = df.groupby('Product_Category')['Sales_Amount'].sum().sort_values(ascending = True)

sns.barplot(
    x= category_sales.values,
    y=category_sales.index,
    color = "orange"
)

plt.title("sales sum according to the product")
plt.xlabel("sales sum")
plt.ylabel("product types")


plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,8))

category_sales = df.groupby('Product_Category')['Sales_Amount'].sum().sort_values(ascending = True)

sns.barplot(
    x= category_sales.values,
    y=category_sales.index,
    color = "pink"
)

plt.title("sales sum according to the product")
plt.xlabel("sales sum")
plt.ylabel("product types")


plt.tight_layout()
plt.show()
```
